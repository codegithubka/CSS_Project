```python
DATA_ROOT = Path.home() / "CSS_Project" / "data"


def load_jsonl(filepath):
    """Load JSONL file into list of dicts."""
    results = []
    with open(filepath, "r") as f:
        for line in f:
            results.append(json.loads(line.strip()))
    return results


def inspect_phase(phase_dir):
    """Inspect a phase's results file."""
    results_file = list(phase_dir.glob("*results*.jsonl"))
    if not results_file:
        print(f"No results file found in {phase_dir}")
        return None

    results_file = results_file[0]
    print(f"\n{'='*60}")
    print(f"Loading: {results_file.name}")
    print(f"From: {phase_dir.name}")

    data = load_jsonl(results_file)
    print(f"Number of records: {len(data)}")

    print(f"\nColumns:")
    for key in sorted(data[0].keys()):
        val = data[0][key]
        dtype = type(val).__name__
        if isinstance(val, list):
            sample = f"list[{len(val)}]" if len(val) > 3 else val
        else:
            sample = val
        print(f"  {key:35} ({dtype:6}): {sample}")

    return data


# Inspect all phases
phases = sorted(DATA_ROOT.glob("phase*"))
for phase_dir in phases:
    inspect_phase(phase_dir)
```

```python
def detect_hydra_effect(prey_death, prey_density):
    """Detect if curve has hydra effect (increasing region).
    
    Parameters
    ----------
    prey_death : array
        Prey death rate values
    prey_density : array
        Corresponding prey density values
        
    Returns
    -------
    has_hydra : bool
        True if curve shows hydra effect (positive derivative region)
    max_deriv : float or None
        Maximum positive derivative if hydra effect detected
    """
    if len(prey_death) < 2:
        return False, None

    derivatives = np.diff(prey_density) / np.diff(prey_death)
    has_increasing = np.any(derivatives > 0)

    if has_increasing:
        max_deriv = np.max(derivatives)
        return True, max_deriv
    return False, None
```

```python
def get_cluster_distribution(df, prey_death_val, species="prey"):
    """Extract all cluster sizes for a given prey_death value."""
    subset = df[np.isclose(df["prey_death"], prey_death_val, atol=0.001)]
    col = f"{species}_cluster_sizes"
    all_sizes = []
    for sizes in subset[col]:
        if isinstance(sizes, list) and len(sizes) > 0:
            all_sizes.extend(sizes)
    return np.array(all_sizes)


# === Plot all cluster sizes as individual points ===


def compute_pdf_all_points(cluster_sizes):
    """Compute P(s) for each unique cluster size."""
    clusters = np.array(cluster_sizes)
    if len(clusters) == 0:
        return np.array([]), np.array([])

    # Count occurrences of each size
    unique, counts = np.unique(clusters, return_counts=True)
    # Normalize to get probability
    probs = counts / len(clusters)

    return unique, probs
```

```python
# =============================================================================
# HYDRA EFFECT ANALYSIS UTILITIES
# =============================================================================
from numpy.polynomial import Polynomial
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.gridspec import GridSpec


def build_curves(df_avg):
    """Build prey density curves for each parameter combination.
    
    Parameters
    ----------
    df_avg : pd.DataFrame
        Averaged data with columns: prey_birth, prey_death, predator_birth, 
        predator_death, prey_density
    
    Returns
    -------
    list of dict
        Each dict contains parameter values and arrays for prey_death/prey_density
    """
    curves_data = []
    for (pb, pred_b, pred_d), group in df_avg.groupby(
        ["prey_birth", "predator_birth", "predator_death"]
    ):
        group_sorted = group.sort_values("prey_death")
        curves_data.append({
            "prey_birth": pb,
            "predator_birth": pred_b,
            "predator_death": pred_d,
            "prey_death": group_sorted["prey_death"].values,
            "prey_density": group_sorted["prey_density"].values,
        })
    return curves_data


def run_hydra_analysis(curves_data):
    """Run full Hydra effect analysis pipeline.
    
    Returns
    -------
    df_quadratic_valid : pd.DataFrame
        Valid quadratic fits with coefficients
    curves_with_hydra : list
        Curves showing Hydra effect
    filtered_curves : list
        Curves truncated at max derivative
    """
    # Detect Hydra effect
    curves_with_hydra = []
    curves_without_hydra = []
    
    for curve in curves_data:
        has_hydra, max_deriv = detect_hydra_effect(
            curve["prey_death"], curve["prey_density"]
        )
        curve["has_hydra"] = has_hydra
        curve["max_derivative"] = max_deriv
        
        if has_hydra:
            curves_with_hydra.append(curve)
        else:
            curves_without_hydra.append(curve)
    
    # Filter curves: keep only portion before max derivative
    filtered_curves = []
    for curve in curves_with_hydra:
        prey_d = curve["prey_death"]
        prey_n = curve["prey_density"]
        
        derivatives = np.diff(prey_n) / np.diff(prey_d)
        max_deriv_idx = np.argmax(derivatives)
        end_idx = max_deriv_idx + 1
        
        if end_idx > 0:
            filtered_curves.append({
                "prey_birth": curve["prey_birth"],
                "predator_birth": curve["predator_birth"],
                "predator_death": curve["predator_death"],
                "prey_death": prey_d[:end_idx + 1],
                "prey_density": prey_n[:end_idx + 1],
                "max_derivative": curve["max_derivative"],
            })
    
    # Fit quadratic polynomial
    curves_with_quadratic = []
    for curve in filtered_curves:
        prey_d = curve["prey_death"]
        prey_n = curve["prey_density"]
        
        if len(prey_d) < 3:
            continue
        
        try:
            poly = Polynomial.fit(prey_d, prey_n, deg=2)
            coeffs = poly.convert().coef
            quadratic_coef = coeffs[2] if len(coeffs) > 2 else 0.0
            
            curves_with_quadratic.append({
                "prey_birth": curve["prey_birth"],
                "predator_birth": curve["predator_birth"],
                "predator_death": curve["predator_death"],
                "quadratic_coef": quadratic_coef,
                "max_derivative": curve["max_derivative"],
            })
        except:
            pass
    
    df_quadratic = pd.DataFrame(curves_with_quadratic)
    df_quadratic_valid = df_quadratic[df_quadratic["quadratic_coef"].notna()]
    
    return df_quadratic_valid, curves_with_hydra, filtered_curves


def plot_hydra_curves(curves_with_hydra, filtered_curves, save_path=None):
    """Plot Hydra effect curves: full and truncated versions."""
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Color normalization
    max_derivs = [c["max_derivative"] for c in curves_with_hydra]
    norm = plt.Normalize(vmin=min(max_derivs), vmax=max(max_derivs))
    cmap = plt.cm.Blues
    
    # Left: All curves with Hydra effect
    ax1 = axes[0]
    for curve in curves_with_hydra:
        color = cmap(norm(curve["max_derivative"]))
        ax1.plot(curve["prey_death"], curve["prey_density"], 
                 color=color, alpha=0.6, linewidth=1.2)
    
    ax1.axhline(0, color="0.5", linestyle="--", linewidth=0.8)
    ax1.set_xlabel("Prey Death Rate")
    ax1.set_ylabel("Equilibrium Prey Density")
    ax1.set_title(f"Curves with Hydra Effect")
    
    # Right: Filtered curves
    ax2 = axes[1]
    for curve in filtered_curves:
        color = cmap(norm(curve["max_derivative"]))
        ax2.plot(curve["prey_death"], curve["prey_density"],
                 color=color, alpha=0.6, linewidth=1.2)
    
    ax2.axhline(0, color="0.5", linestyle="--", linewidth=0.8)
    ax2.set_xlabel("Prey Death Rate")
    ax2.set_ylabel("Equilibrium Prey Density")
    ax2.set_title(f"Truncated at Max Derivative")
    
    # Match y-axis limits
    y_max = max(ax1.get_ylim()[1], ax2.get_ylim()[1])
    y_min = min(ax1.get_ylim()[0], ax2.get_ylim()[0])
    ax1.set_ylim(y_min, y_max)
    ax2.set_ylim(y_min, y_max)
    
    # Shared colorbar
    fig.subplots_adjust(right=0.88)
    cbar_ax = fig.add_axes([0.90, 0.15, 0.02, 0.7])
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = fig.colorbar(sm, cax=cbar_ax)
    cbar.set_label("Max Positive Derivative")
    
    fig.suptitle("Hydra Effect: Full vs Truncated Curves", y=1.02, fontweight="normal")
    plt.tight_layout(rect=[0, 0, 0.88, 1])
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches="tight")
    plt.show()


def plot_3d_heatmap(df_quadratic_valid, title_suffix=""):
    """Generate 3D voxel heatmap of quadratic coefficients."""
    if len(df_quadratic_valid) == 0:
        print("No valid data for 3D heatmap")
        return
    
    # Get unique parameter values
    pb_grid = np.array(sorted(df_quadratic_valid["prey_birth"].unique()))
    pred_b_grid = np.array(sorted(df_quadratic_valid["predator_birth"].unique()))
    pred_d_grid = np.array(sorted(df_quadratic_valid["predator_death"].unique()))
    
    # Create 3D array
    heatmap_3d = np.full((len(pb_grid), len(pred_b_grid), len(pred_d_grid)), np.nan)
    
    for _, row in df_quadratic_valid.iterrows():
        pb_idx = np.where(np.isclose(pb_grid, row["prey_birth"]))[0]
        pred_b_idx = np.where(np.isclose(pred_b_grid, row["predator_birth"]))[0]
        pred_d_idx = np.where(np.isclose(pred_d_grid, row["predator_death"]))[0]
        
        if len(pb_idx) > 0 and len(pred_b_idx) > 0 and len(pred_d_idx) > 0:
            heatmap_3d[pb_idx[0], pred_b_idx[0], pred_d_idx[0]] = row["quadratic_coef"]
    
    # Color normalization
    vmin = np.nanpercentile(heatmap_3d, 5)
    vmax = np.nanpercentile(heatmap_3d, 95)
    norm = plt.Normalize(vmin=vmin, vmax=vmax)
    cmap = plt.cm.Blues
    
    # Create voxel colors
    filled = ~np.isnan(heatmap_3d)
    colors = np.empty(heatmap_3d.shape + (4,))
    
    for i in range(heatmap_3d.shape[0]):
        for j in range(heatmap_3d.shape[1]):
            for k in range(heatmap_3d.shape[2]):
                if filled[i, j, k]:
                    rgb = cmap(norm(heatmap_3d[i, j, k]))[:3]
                    colors[i, j, k] = [rgb[0], rgb[1], rgb[2], 0.9]
                else:
                    colors[i, j, k] = [0, 0, 0, 0]
    
    # Create figure
    fig = plt.figure(figsize=(16, 7))
    gs = GridSpec(1, 3, width_ratios=[1, 1, 0.05], wspace=0.25)
    
    tick_skip = 2
    
    # Left panel: Voxel plot
    ax1 = fig.add_subplot(gs[0, 0], projection="3d")
    ax1.voxels(filled, facecolors=colors, edgecolors="0.4", linewidth=0.1)
    
    ax1.set_xticks(range(0, len(pb_grid), tick_skip))
    ax1.set_xticklabels([f"{pb_grid[i]:.1f}" for i in range(0, len(pb_grid), tick_skip)])
    ax1.set_yticks(range(0, len(pred_b_grid), tick_skip))
    ax1.set_yticklabels([f"{pred_b_grid[i]:.1f}" for i in range(0, len(pred_b_grid), tick_skip)])
    ax1.set_zticks(range(0, len(pred_d_grid), tick_skip))
    ax1.set_zticklabels([f"{pred_d_grid[i]:.1f}" for i in range(0, len(pred_d_grid), tick_skip)])
    
    ax1.set_xlabel("Prey Birth")
    ax1.set_ylabel("Predator Birth")
    ax1.set_zlabel("Predator Death")
    ax1.set_title("3D Voxel Heatmap")
    ax1.view_init(elev=25, azim=225)
    
    # Right panel: Cross-sections
    ax2 = fig.add_subplot(gs[0, 1], projection="3d")
    PB_slice, PRED_B_slice = np.meshgrid(pb_grid, pred_b_grid, indexing="ij")
    
    n_slices = min(5, len(pred_d_grid))
    slice_indices = np.linspace(0, len(pred_d_grid) - 1, n_slices, dtype=int)
    
    for slice_idx in slice_indices:
        pd_val = pred_d_grid[slice_idx]
        slice_data = heatmap_3d[:, :, slice_idx]
        valid_mask = ~np.isnan(slice_data)
        
        if not np.any(valid_mask):
            continue
        
        Z_slice = np.full_like(PB_slice, pd_val)
        face_colors = np.empty((*slice_data.shape, 4))
        
        for i in range(slice_data.shape[0]):
            for j in range(slice_data.shape[1]):
                if valid_mask[i, j]:
                    face_colors[i, j] = [*cmap(norm(slice_data[i, j]))[:3], 0.8]
                else:
                    face_colors[i, j] = [0, 0, 0, 0]
        
        PB_masked = np.ma.masked_where(~valid_mask, PB_slice)
        PRED_B_masked = np.ma.masked_where(~valid_mask, PRED_B_slice)
        Z_masked = np.ma.masked_where(~valid_mask, Z_slice)
        
        ax2.plot_surface(PB_masked, PRED_B_masked, Z_masked,
                         facecolors=face_colors, alpha=0.8, linewidth=0.2, shade=False)
    
    ax2.set_xlim(pb_grid.min(), pb_grid.max())
    ax2.set_ylim(pred_b_grid.min(), pred_b_grid.max())
    ax2.set_zlim(pred_d_grid.min(), pred_d_grid.max())
    ax2.set_xticks(pb_grid[::tick_skip])
    ax2.set_xticklabels([f"{x:.1f}" for x in pb_grid[::tick_skip]])
    ax2.set_yticks(pred_b_grid[::tick_skip])
    ax2.set_yticklabels([f"{x:.1f}" for x in pred_b_grid[::tick_skip]])
    ax2.set_zticks(pred_d_grid[::tick_skip])
    ax2.set_zticklabels([f"{x:.1f}" for x in pred_d_grid[::tick_skip]])
    
    ax2.set_xlabel("Prey Birth")
    ax2.set_ylabel("Predator Birth")
    ax2.set_zlabel("Predator Death")
    ax2.set_title("Cross-Sections")
    ax2.view_init(elev=25, azim=225)
    
    # Colorbar
    cbar_ax = fig.add_subplot(gs[0, 2])
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = fig.colorbar(sm, cax=cbar_ax)
    cbar.set_label("Quadratic Coefficient")
    
    fig.suptitle(f"Hydra Effect Strength{title_suffix}: 3D Parameter Space", 
                 y=0.98, fontweight="normal")
    plt.savefig("phase4_3d_heatmap.png", dpi=150, bbox_inches="tight")
    plt.show()
    
    print(f"Data coverage: {100*np.sum(filled)/filled.size:.1f}%")
    print(f"Coefficient range: [{vmin:.1f}, {vmax:.1f}]")


def plot_marginal_effects(df_quadratic_valid):
    """Plot marginal effects of each parameter on Hydra strength."""
    fig, axes = plt.subplots(1, 3, figsize=(14, 4))
    
    # Plot 1: Effect of Prey Birth
    by_pb = (df_quadratic_valid.groupby("prey_birth")["quadratic_coef"]
             .agg(["mean", "std", "count"]).reset_index())
    by_pb["se"] = by_pb["std"] / np.sqrt(by_pb["count"])
    
    ax1 = axes[0]
    ax1.errorbar(by_pb["prey_birth"], by_pb["mean"], yerr=by_pb["se"],
                 fmt="o-", color=COLORS["prey"], markersize=6, linewidth=1.5, capsize=3)
    ax1.axhline(0, color="0.5", linestyle="--", linewidth=1)
    ax1.set_xlabel("Prey Birth Rate")
    ax1.set_ylabel("Quadratic Coefficient")
    ax1.set_title("Effect of Prey Birth")
    ax1.fill_between(by_pb["prey_birth"], 0, by_pb["mean"], alpha=0.1, color=COLORS["prey"])
    
    # Plot 2: Effect of Predator Birth
    by_pred_b = (df_quadratic_valid.groupby("predator_birth")["quadratic_coef"]
                 .agg(["mean", "std", "count"]).reset_index())
    by_pred_b["se"] = by_pred_b["std"] / np.sqrt(by_pred_b["count"])
    
    ax2 = axes[1]
    ax2.errorbar(by_pred_b["predator_birth"], by_pred_b["mean"], yerr=by_pred_b["se"],
                 fmt="s-", color=COLORS["predator"], markersize=6, linewidth=1.5, capsize=3)
    ax2.axhline(0, color="0.5", linestyle="--", linewidth=1)
    ax2.set_xlabel("Predator Birth Rate")
    ax2.set_ylabel("Quadratic Coefficient")
    ax2.set_title("Effect of Predator Birth")
    ax2.fill_between(by_pred_b["predator_birth"], 0, by_pred_b["mean"], 
                     alpha=0.1, color=COLORS["predator"])
    
    # Plot 3: Effect of Predator Death
    by_pred_d = (df_quadratic_valid.groupby("predator_death")["quadratic_coef"]
                 .agg(["mean", "std", "count"]).reset_index())
    by_pred_d["se"] = by_pred_d["std"] / np.sqrt(by_pred_d["count"])
    
    ax3 = axes[2]
    ax3.errorbar(by_pred_d["predator_death"], by_pred_d["mean"], yerr=by_pred_d["se"],
                 fmt="^-", color=COLORS["highlight"], markersize=6, linewidth=1.5, capsize=3)
    ax3.axhline(0, color="0.5", linestyle="--", linewidth=1)
    ax3.set_xlabel("Predator Death Rate")
    ax3.set_ylabel("Quadratic Coefficient")
    ax3.set_title("Effect of Predator Death")
    ax3.fill_between(by_pred_d["predator_death"], 0, by_pred_d["mean"],
                     alpha=0.1, color=COLORS["highlight"])
    
    fig.suptitle("Hydra Effect Strength: Marginal Effects", y=1.02)
    plt.tight_layout()
    plt.show()


def plot_validation_curves(df_quadratic_valid, df_avg, title_suffix=""):
    """Plot validation: actual curves for strong vs moderate Hydra effect."""
    strong_hydra = df_quadratic_valid[df_quadratic_valid["quadratic_coef"] > 15]
    moderate_hydra = df_quadratic_valid[
        (df_quadratic_valid["quadratic_coef"] >= 5) & 
        (df_quadratic_valid["quadratic_coef"] <= 10)
    ]
    
    print(f"Strong Hydra (coef > 15): {len(strong_hydra)} curves")
    print(f"Moderate Hydra (5 ≤ coef ≤ 10): {len(moderate_hydra)} curves")
    
    n_samples = min(8, len(strong_hydra))
    n_samples_mod = min(8, len(moderate_hydra))
    
    strong_sample = strong_hydra.nlargest(n_samples, "quadratic_coef")
    moderate_sample = (moderate_hydra.sample(n_samples_mod, random_state=42)
                       if len(moderate_hydra) >= n_samples_mod else moderate_hydra)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Left: Strong Hydra Effect
    ax1 = axes[0]
    colors_strong = plt.cm.Blues(np.linspace(0.4, 0.9, len(strong_sample)))
    
    for i, (_, row) in enumerate(strong_sample.iterrows()):
        mask = ((df_avg["prey_birth"] == row["prey_birth"]) &
                (df_avg["predator_birth"] == row["predator_birth"]) &
                (df_avg["predator_death"] == row["predator_death"]))
        curve_data = df_avg[mask].sort_values("prey_death")
        
        if len(curve_data) > 0:
            ax1.plot(curve_data["prey_death"], curve_data["prey_density"], "-",
                     color=colors_strong[i], linewidth=1.5, alpha=0.8,
                     label=f'c={row["quadratic_coef"]:.1f}')
    
    ax1.set_xlabel("Prey Death Rate")
    ax1.set_ylabel("Equilibrium Prey Density")
    ax1.set_title(f"Strong Hydra Effect (coef > 15, n={len(strong_sample)})")
    ax1.legend(loc="upper right", fontsize=8, framealpha=0.95, title="Quad. Coef.")
    ax1.axhline(0, color="0.5", linestyle="--", linewidth=0.8)
    
    # Right: Moderate Hydra Effect
    ax2 = axes[1]
    colors_moderate = plt.cm.Oranges(np.linspace(0.4, 0.9, len(moderate_sample)))
    
    for i, (_, row) in enumerate(moderate_sample.iterrows()):
        mask = ((df_avg["prey_birth"] == row["prey_birth"]) &
                (df_avg["predator_birth"] == row["predator_birth"]) &
                (df_avg["predator_death"] == row["predator_death"]))
        curve_data = df_avg[mask].sort_values("prey_death")
        
        if len(curve_data) > 0:
            ax2.plot(curve_data["prey_death"], curve_data["prey_density"], "-",
                     color=colors_moderate[i], linewidth=1.5, alpha=0.8,
                     label=f'c={row["quadratic_coef"]:.1f}')
    
    ax2.set_xlabel("Prey Death Rate")
    ax2.set_ylabel("Equilibrium Prey Density")
    ax2.set_title(f"Moderate Hydra Effect (5 ≤ coef ≤ 10, n={len(moderate_sample)})")
    ax2.legend(loc="upper right", fontsize=8, framealpha=0.95, title="Quad. Coef.")
    ax2.axhline(0, color="0.5", linestyle="--", linewidth=0.8)
    
    # Match y-axis limits
    y_max = max(ax1.get_ylim()[1], ax2.get_ylim()[1])
    ax1.set_ylim(bottom=0, top=y_max)
    ax2.set_ylim(bottom=0, top=y_max)
    
    fig.suptitle(f"Quadratic Coefficient Captures Hydra Effect Strength{title_suffix}",
                 y=1.02, fontweight="normal")
    plt.tight_layout()
    plt.show()
    
    # Print parameter details
    print("\n=== Strong Hydra Parameters ===")
    print(strong_sample[["prey_birth", "predator_birth", "predator_death", 
                         "quadratic_coef"]].to_string(index=False))
    print("\n=== Moderate Hydra Parameters ===")
    print(moderate_sample[["prey_birth", "predator_birth", "predator_death",
                           "quadratic_coef"]].to_string(index=False))
```

```python
# === Critical Point Cluster Distribution Plot ===

# Get clusters at critical point
clusters_crit = get_cluster_distribution(df_phase1, 0.095, "prey")

print(f"Clusters at d ≈ 0.095: {len(clusters_crit)}")
print(f"Size range: [{clusters_crit.min()}, {clusters_crit.max()}]")

# Compute PDF (all points)
x_crit, y_crit = compute_pdf_all_points(clusters_crit)

# Fit power law
fit_crit = powerlaw.Fit(clusters_crit, discrete=True, verbose=False)

# Distribution comparisons
R_pl_tpl, p_pl_tpl = fit_crit.distribution_compare(
    "power_law", "truncated_power_law", normalized_ratio=True
)
R_pl_ln, p_pl_ln = fit_crit.distribution_compare(
    "power_law", "lognormal", normalized_ratio=True
)
R_tpl_ln, p_tpl_ln = fit_crit.distribution_compare(
    "truncated_power_law", "lognormal", normalized_ratio=True
)

print("=== Power Law Fit (Critical Point, d ≈ 0.095) ===")
print(f"Power law α: {fit_crit.power_law.alpha:.3f}")
print(f"xmin: {fit_crit.power_law.xmin}")
print(f"Truncated PL α: {fit_crit.truncated_power_law.alpha:.3f}")
print(f"Truncated PL λ: {fit_crit.truncated_power_law.parameter2:.6f}")

print(f"\n=== Distribution Comparison ===")
print(f"Power law vs Truncated PL:      R={R_pl_tpl:+.3f}, p={p_pl_tpl:.4f}")
print(f"Power law vs Lognormal:         R={R_pl_ln:+.3f}, p={p_pl_ln:.4f}")
print(f"Truncated PL vs Lognormal:      R={R_tpl_ln:+.3f}, p={p_tpl_ln:.4f}")
```

```python
# === Predator α Across All Prey Death Values ===

prey_deaths_p1 = sorted(df_phase1["prey_death"].unique())

results = []
for pd_val in prey_deaths_p1:
    pred_clusters = get_cluster_distribution(df_phase1, pd_val, "pred")

    if len(pred_clusters) < 100:
        continue

    fit = powerlaw.Fit(pred_clusters, discrete=True, xmin=1, verbose=False)
    R_tpl_ln, _ = fit.distribution_compare(
        "truncated_power_law", "lognormal", normalized_ratio=True
    )

    results.append(
        {
            "prey_death": pd_val,
            "n_clusters": len(pred_clusters),
            "max_size": pred_clusters.max(),
            "alpha": fit.truncated_power_law.alpha,
            "R_tpl_ln": R_tpl_ln,
        }
    )

results_df = pd.DataFrame(results)

# === Plot ===
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Left: α vs prey_death
ax1 = axes[0]
ax1.plot(
    results_df["prey_death"],
    results_df["alpha"],
    "o-",
    color=COLORS["predator"],
    markersize=7,
    linewidth=1.5,
)
ax1.axhline(
    1.17, color="0.5", linestyle="--", linewidth=1, label="α = 1.17 (reference)"
)
ax1.set_xlabel("Prey Death Rate")
ax1.set_ylabel("Exponent α")
ax1.set_title("Predator Power Law Exponent")
ax1.legend(loc="upper right", framealpha=0.95)

# Right: R vs prey_death
ax2 = axes[1]
bars = ax2.bar(
    results_df["prey_death"],
    results_df["R_tpl_ln"],
    width=0.008,
    color=COLORS["predator"],
    alpha=0.8,
    edgecolor="none",
)
ax2.axhline(0, color="0.3", linewidth=1)
ax2.set_xlabel("Prey Death Rate")
ax2.set_ylabel("R (TPL vs Lognormal)")
ax2.set_title("Power Law Fit Quality (R > 0 = TPL wins)")

# Add annotation
ax2.text(
    0.95,
    0.95,
    "R > 0 everywhere\n→ Always power law",
    transform=ax2.transAxes,
    ha="right",
    va="top",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="white", edgecolor="0.8", alpha=0.9),
)

plt.tight_layout()
plt.savefig("plot_predator_alpha_sweep.png", dpi=150, bbox_inches="tight")
plt.show()
```

NOTE:

Predator clusters follow a truncated-power law distrbutions across all parameter values where predators survive. The exponent decreases as we go toward extinction and maximum cluster size collapses. This would suggest that movement dynamics generate scale-free spatial patterns, independent of proximity to any critical point.

```python
# === Phase 3: Finite-Size Scaling - Lognormal Fit ===
from scipy import stats

# Load Phase 3 data
df_phase3 = load_phase_to_df(DATA_ROOT / "phase3_18698382")

print(f"Phase 3 loaded: {len(df_phase3)} records")
print(f"Grid sizes: {sorted(df_phase3['grid_size'].unique())}")
print(f"prey_death: {df_phase3['prey_death'].unique()}")

grid_sizes = sorted(df_phase3["grid_size"].unique())

# Color palette for different L values
colors_L = plt.cm.viridis(np.linspace(0, 0.9, len(grid_sizes)))

# Store fit results
fit_results = []

for i, L in enumerate(grid_sizes):
    subset = df_phase3[df_phase3["grid_size"] == L]

    clusters = []
    for sizes in subset["prey_cluster_sizes"]:
        if isinstance(sizes, list) and len(sizes) > 0:
            clusters.extend(sizes)

    if len(clusters) < 100:
        continue

    clusters = np.array(clusters)

    # Fit lognormal using scipy
    shape, loc, scale = stats.lognorm.fit(clusters, floc=0)
    mu = np.log(scale)
    sigma = shape

    # Statistical test: power law vs lognormal
    fit = powerlaw.Fit(clusters, discrete=True, xmin=1, verbose=False)
    R_pl_ln, _ = fit.distribution_compare(
        "power_law", "lognormal", normalized_ratio=True
    )

    # Compute PDF
    unique, counts = np.unique(clusters, return_counts=True)
    probs = counts / len(clusters)

    fit_results.append(
        {
            "L": L,
            "unique": unique,
            "probs": probs,
            "mu": mu,
            "sigma": sigma,
            "R_pl_ln": R_pl_ln,
            "color": colors_L[i],
        }
    )
```

```python
# === Prey Cluster FSS with Lognormal Fits ===

fig, ax = plt.subplots(figsize=(9, 6))

for res in fit_results:
    L = res["L"]
    ax.scatter(
        res["unique"],
        res["probs"],
        color=res["color"],
        s=12,
        alpha=0.6,
        edgecolors="none",
        label=f'L={L} (μ={res["mu"]:.2f}, σ={res["sigma"]:.2f}, R={res["R_pl_ln"]:+.1f})',
    )

ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Cluster Size")
ax.set_ylabel("P(s)")
ax.set_title("Prey Cluster Distribution: Finite-Size Scaling (d = 0.0955)")

# Annotation instead of subtitle
ax.text(
    0.02,
    0.02,
    "R < 0: Lognormal",
    transform=ax.transAxes,
    ha="left",
    va="bottom",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="white", edgecolor="0.8", alpha=0.9),
)

# Legend at bottom
ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.10),
    ncol=2,
    fontsize=9,
    framealpha=0.95,
    edgecolor="0.8",
)

plt.tight_layout()
plt.subplots_adjust(bottom=0.20)
plt.savefig("plot_prey_fss_lognormal.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# === Prey vs Predator Distribution Comparison ===
from scipy import stats

prey_deaths_p1 = sorted(df_phase1["prey_death"].unique())

# Collect prey data
results_prey = []
for pd_val in prey_deaths_p1:
    if pd_val > 0.10:  # Skip noisy collapse region
        continue

    prey_clusters = get_cluster_distribution(df_phase1, pd_val, "prey")

    if len(prey_clusters) < 100:
        continue

    fit = powerlaw.Fit(prey_clusters, discrete=True, xmin=1, verbose=False)
    R_tpl_ln, _ = fit.distribution_compare(
        "truncated_power_law", "lognormal", normalized_ratio=True
    )
    shape, loc, scale = stats.lognorm.fit(prey_clusters, floc=0)

    results_prey.append(
        {
            "prey_death": pd_val,
            "n_clusters": len(prey_clusters),
            "max_size": prey_clusters.max(),
            "alpha": fit.truncated_power_law.alpha,
            "R_tpl_ln": R_tpl_ln,
            "mu": np.log(scale),
            "sigma": shape,
        }
    )

results_prey_df = pd.DataFrame(results_prey)

# Collect predator data
results_pred = []
for pd_val in prey_deaths_p1:
    pred_clusters = get_cluster_distribution(df_phase1, pd_val, "pred")

    if len(pred_clusters) < 100:
        continue

    fit = powerlaw.Fit(pred_clusters, discrete=True, xmin=1, verbose=False)
    R_tpl_ln, _ = fit.distribution_compare(
        "truncated_power_law", "lognormal", normalized_ratio=True
    )

    results_pred.append(
        {
            "prey_death": pd_val,
            "n_clusters": len(pred_clusters),
            "max_size": pred_clusters.max(),
            "alpha": fit.truncated_power_law.alpha,
            "R_tpl_ln": R_tpl_ln,
        }
    )

results_pred_df = pd.DataFrame(results_pred)

# === Plot ===
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# Top Left: Prey R
ax1 = axes[0, 0]
ax1.bar(
    results_prey_df["prey_death"],
    results_prey_df["R_tpl_ln"],
    width=0.008,
    color=COLORS["prey"],
    alpha=0.8,
    edgecolor="none",
)
ax1.axhline(0, color="0.3", linewidth=1)
ax1.set_xlabel("Prey Death Rate")
ax1.set_ylabel("R (TPL vs Lognormal)")
ax1.set_title("Prey: Lognormal Wins (R < 0)")
ax1.text(
    0.95,
    0.95,
    "R < 0 everywhere",
    transform=ax1.transAxes,
    ha="right",
    va="top",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="white", edgecolor="0.8", alpha=0.9),
)

# Top Right: Predator R
ax2 = axes[0, 1]
ax2.bar(
    results_pred_df["prey_death"],
    results_pred_df["R_tpl_ln"],
    width=0.008,
    color=COLORS["predator"],
    alpha=0.8,
    edgecolor="none",
)
ax2.axhline(0, color="0.3", linewidth=1)
ax2.set_xlabel("Prey Death Rate")
ax2.set_ylabel("R (TPL vs Lognormal)")
ax2.set_title("Predator: Power Law Wins (R > 0)")
ax2.text(
    0.95,
    0.95,
    "R > 0 everywhere",
    transform=ax2.transAxes,
    ha="right",
    va="top",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="white", edgecolor="0.8", alpha=0.9),
)

# Bottom Left: Prey lognormal params
ax3 = axes[1, 0]
ax3.plot(
    results_prey_df["prey_death"],
    results_prey_df["mu"],
    "o-",
    color=COLORS["prey"],
    markersize=7,
    linewidth=1.5,
    label="μ",
)
ax3.plot(
    results_prey_df["prey_death"],
    results_prey_df["sigma"],
    "s--",
    color=COLORS["prey"],
    markersize=7,
    linewidth=1.5,
    alpha=0.6,
    label="σ",
)
ax3.set_xlabel("Prey Death Rate")
ax3.set_ylabel("Lognormal Parameters")
ax3.set_title("Prey Lognormal Parameters")
ax3.legend(loc="upper right", framealpha=0.95)

# Add summary stats as annotation
mu_mean, mu_std = results_prey_df["mu"].mean(), results_prey_df["mu"].std()
sigma_mean, sigma_std = results_prey_df["sigma"].mean(), results_prey_df["sigma"].std()
ax3.text(
    0.02,
    0.02,
    f"μ = {mu_mean:.2f} ± {mu_std:.2f}\nσ = {sigma_mean:.2f} ± {sigma_std:.2f}",
    transform=ax3.transAxes,
    ha="left",
    va="bottom",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="white", edgecolor="0.8", alpha=0.9),
)

# Bottom Right: Predator α
ax4 = axes[1, 1]
ax4.plot(
    results_pred_df["prey_death"],
    results_pred_df["alpha"],
    "o-",
    color=COLORS["predator"],
    markersize=7,
    linewidth=1.5,
)
ax4.set_xlabel("Prey Death Rate")
ax4.set_ylabel("Exponent α")
ax4.set_title("Predator Power Law Exponent")

# Add range annotation
alpha_min, alpha_max = results_pred_df["alpha"].min(), results_pred_df["alpha"].max()
ax4.text(
    0.02,
    0.02,
    f"α: {alpha_min:.2f} → {alpha_max:.2f}",
    transform=ax4.transAxes,
    ha="left",
    va="bottom",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="white", edgecolor="0.8", alpha=0.9),
)

plt.tight_layout()
plt.show()
```
